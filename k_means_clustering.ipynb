# K-Means Clustering

## Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```python
dataset = pd.read_csv('data/Mall_Customers.csv')
X = dataset.iloc[:, [3, 4]].values
```

## Using the elbow method to find the optimal number of clusters

```python
from sklearn.cluster import KMeans
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters = i, init = 'k-means++', random_state = 42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
plt.plot(range(1, 11), wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

## Training the K-Means model on the dataset

```python
kmeans = KMeans(n_clusters = 5, init = 'k-means++', random_state = 42)
y_kmeans = kmeans.fit_predict(X)
```

## Visualising the clusters

```python
plt.scatter(X[y_kmeans == 0, 0], X[y_kmeans == 0, 1], s = 100, c = 'red', label = 'Cluster 1')
plt.scatter(X[y_kmeans == 1, 0], X[y_kmeans == 1, 1], s = 100, c = 'blue', label = 'Cluster 2')
plt.scatter(X[y_kmeans == 2, 0], X[y_kmeans == 2, 1], s = 100, c = 'green', label = 'Cluster 3')
plt.scatter(X[y_kmeans == 3, 0], X[y_kmeans == 3, 1], s = 100, c = 'cyan', label = 'Cluster 4')
plt.scatter(X[y_kmeans == 4, 0], X[y_kmeans == 4, 1], s = 100, c = 'magenta', label = 'Cluster 5')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s = 300, c = 'yellow', label = 'Centroids')
plt.title('Clusters of customers')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.show()
```

## Apriori Algorithm


```python
dataset = pd.read_csv('data/Market_Basket_Optimisation.csv', header=None)
```

```python
dataset.head()
```

```python
transactions = []
for i in range(0, 7501):
    transactions.append([str(dataset.values[i, j]) for j in range(0, 20)])
```

```python
transactions
```

```python
from apyori import apriori
```

```python
rules = apriori(transactions, min_support = 0.003, min_confidence = 0.2, min_lift = 3, min_length = 2)
```

```python
results = list(rules)
results
```

```python
def inspect(results):
    lhs         = [tuple(result[2][0][0])[0] for result in results]
    rhs         = [tuple(result[2][0][1])[0] for result in results]
    supports    = [result[1] for result in results]
    confidences = [result[2][0][2] for result in results]
    lifts       = [result[2][0][3] for result in results]
    return list(zip(lhs, rhs, supports, confidences, lifts))
resultsinDataFrame = pd.DataFrame(inspect(results), columns = ['Left Hand Side', 'Right Hand Side', 'Support', 'Confidence', 'Lift'])
```

```python
resultsinDataFrame
```

```python
resultsinDataFrame.sort_values(by = 'Lift', ascending = False).head(10)
```

## DBSCAN

```python
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN     
```

```python
X, _ = make_moons(n_samples=300, noise=0.08, random_state=42)
```

```python
# 2. Always scale your features! 
# Distance metrics like Euclidean distance are sensitive to feature scales.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
# 3. Fit DBSCAN
# eps: neighborhood radius
# min_samples: minimum points required to form a core point
dbscan = DBSCAN(eps=0.3, min_samples=5)
clusters = dbscan.fit_predict(X_scaled)
clusters
```

```python
# 4. Inspect results
# In scikit-learn, noise points are assigned the label -1
n_clusters = len(set(clusters)) - (1 if -1 in clusters else 0)
n_noise = list(clusters).count(-1)
n_clusters, n_noise
```

```python
plt.figure(figsize=(8, 5))
scatter = plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=clusters, cmap='viridis', s=40, edgecolors='k')
plt.title(f"DBSCAN Clustering (Clusters: {n_clusters}, Noise: {n_noise})")
plt.xlabel("Feature 1 (scaled)")
plt.ylabel("Feature 2 (scaled)")
plt.colorbar(scatter, label="Cluster Label (-1 = Noise)")
plt.show()
```

```python
from sklearn.cluster import KMeans

# Fit K-Means on the same scaled data
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans_clusters = kmeans.fit_predict(X_scaled)

# Plot comparison
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot K-Means
axes[0].scatter(X_scaled[:, 0], X_scaled[:, 1], c=kmeans_clusters, cmap='Accent', s=40, edgecolors='k')
axes[0].set_title("K-Means (Fails on non-spherical shapes)")

# Plot DBSCAN
axes[1].scatter(X_scaled[:, 0], X_scaled[:, 1], c=clusters, cmap='viridis', s=40, edgecolors='k')
axes[1].set_title("DBSCAN (Correctly follows density shapes)")

plt.show()
```

## OPTICS Model

```python
from sklearn.cluster import OPTICS
from sklearn.datasets import make_blobs
```

```python
X1, _ = make_blobs(n_samples=150, centers=[[0,0]], cluster_std=0.20, random_state=42)
X2, _ = make_blobs(n_samples=300, centers=[[2,2]], cluster_std=0.80, random_state=42)
X = np.vstack((X1, X2))
X
```

```python
# 2. Fit OPTICS
# min_samples: minimum points in a neighborhood
# xi: determines the steepness threshold to separate clusters
optics_model = OPTICS(min_samples=20, xi=0.05, min_cluster_size=0.1)
optics_model.fit(X)
```

```python
X2
```

```python
# 3. Extract the reachability distances and ordering
labels = optics_model.labels_
space = np.arange(len(X))
reachability = optics_model.reachability_[optics_model.ordering_]
```

```python
# 4. Plot Reachability Diagram
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
# Reachability Plot
ax1.plot(space, reachability, 'b-')
ax1.set_title("OPTICS Reachability Plot")
ax1.set_xlabel("Points Ordered by OPTICS")
ax1.set_ylabel("Reachability Distance")
ax1.grid(True)


# Clustered Data Plot
colors = ['g', 'r', 'b', 'y', 'c']
for klass in set(labels):
    Xk = X[labels == klass]
    if klass == -1:
        # Black for noise
        ax2.scatter(Xk[:, 0], Xk[:, 1], color='k', alpha=0.3, label='Noise')
    else:
        ax2.scatter(Xk[:, 0], Xk[:, 1], label=f'Cluster {klass}')

ax2.set_title("Extracted Clusters (Varying Densities)")
ax2.legend()
plt.show()
```

```python
# Compare OPTICS and DBSCAN on the same dataset

# 1. Fit DBSCAN on the same dataset
dbscan = DBSCAN(eps=0.3, min_samples=20)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_pred = dbscan.fit_predict(X_scaled)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

colors = ['g', 'r', 'b', 'y', 'c']
for klass in set(labels):
    Xk = X[labels == klass]
    if klass == -1:
        # Black for noise
        ax1.scatter(Xk[:, 0], Xk[:, 1], color='k', alpha=0.3, label='Noise')
    else:
        ax1.scatter(Xk[:, 0], Xk[:, 1], label=f'Cluster {klass}')

ax1.set_title("OPTICS Clustering (Varying Densities)")

ax2.scatter(X_scaled[:, 0], X_scaled[:, 1], c=X_pred, cmap='viridis', s=40, edgecolors='k')
ax2.set_title("DBSCAN (Correctly follows density shapes)")

```

```python
from matplotlib.patches import Ellipse
from sklearn.datasets import make_blobs
from sklearn.mixture import GaussianMixture
```

```python
# 1. Create synthetic data: 3 stretched / rotated elliptical clusters
X, _ = make_blobs(n_samples=600, centers=3, cluster_std=0.60, random_state=42)
```

```python
# Apply a transformation matrix to stretch and tilt the data
transformation = [[0.6, -0.6], [-0.4, 0.8]]
X_stretched = np.dot(X, transformation)
```

```python
# 2. Fit the Gaussian Mixture Model
# n_components: number of Gaussian bell curves (clusters)
# covariance_type='full': allows ellipses to tilt and stretch in any direction
gmm = GaussianMixture(n_components=3, covariance_type='full', random_state=42)
gmm.fit(X_stretched)
```

```python
# 3. Hard vs. Soft Predictions
hard_clusters = gmm.predict(X_stretched)                  # Assigns 0, 1, or 2
soft_probabilities = gmm.predict_proba(X_stretched)      # Probabilities [p0, p1, p2]
```

```python
# Let's inspect the soft probabilities for the first 3 data points:
print("--- Soft Probabilities for First 3 Points ---")
for i in range(3):
    probs = [f"{p*100:.1f}%" for p in soft_probabilities[i]]
    print(f"Point {i}: Cluster 0: {probs[0]}, Cluster 1: {probs[1]}, Cluster 2: {probs[2]}")
```

```python
# 4. Helper function to draw the Gaussian ellipses (confidence zones)
def draw_ellipse(position, covariance, ax, **kwargs):
    """Draws an ellipse representing 2 standard deviations (~95% of data)."""
    # Calculate eigenvalues and eigenvectors of covariance matrix
    eigenvalues, eigenvectors = np.linalg.eigh(covariance)
    order = eigenvalues.argsort()[::-1]
    eigenvalues, eigenvectors = eigenvalues[order], eigenvectors[:, order]
    
    # Calculate orientation angle in degrees
    angle = np.degrees(np.arctan2(*eigenvectors[:, 0][::-1]))
    
    # 2 standard deviations covers ~95.4% of a normal distribution
    width, height = 2 * 2 * np.sqrt(eigenvalues)
    
    ellipse = Ellipse(xy=position, width=width, height=height, angle=angle, **kwargs)
    ax.add_patch(ellipse)
```

```python
# 5. Visualize the Clusters & Learned Gaussians
fig, ax = plt.subplots(figsize=(9, 6))

# Plot the scatter points colored by hard cluster assignment
scatter = ax.scatter(
    X_stretched[:, 0], 
    X_stretched[:, 1], 
    c=hard_clusters, 
    cmap='viridis', 
    s=30, 
    alpha=0.6, 
    edgecolors='k'
)

# Plot the center (mean) and outer boundary of each Gaussian
for i in range(gmm.n_components):
    mean = gmm.means_[i]
    covariance = gmm.covariances_[i]
    
    # Draw center point (peak of the bell curve)
    ax.scatter(mean[0], mean[1], c='red', s=120, marker='X', zorder=10)
    
    # Draw 95% confidence ellipse
    draw_ellipse(mean, covariance, ax, color='red', fill=False, linewidth=2.5, linestyle='--')

ax.set_title("GMM: Elliptical Clusters with 95% Confidence Boundaries")
ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")
plt.show()
```
