# K-Means Clustering

## Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```python
dataset = pd.read_csv('data/Mall_Customers.csv')
X = dataset.iloc[:, [3, 4]].values
```

## Using the elbow method to find the optimal number of clusters

```python
from sklearn.cluster import KMeans
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters = i, init = 'k-means++', random_state = 42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
plt.plot(range(1, 11), wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

## Training the K-Means model on the dataset

```python
kmeans = KMeans(n_clusters = 5, init = 'k-means++', random_state = 42)
y_kmeans = kmeans.fit_predict(X)
```

## Visualising the clusters

```python
plt.scatter(X[y_kmeans == 0, 0], X[y_kmeans == 0, 1], s = 100, c = 'red', label = 'Cluster 1')
plt.scatter(X[y_kmeans == 1, 0], X[y_kmeans == 1, 1], s = 100, c = 'blue', label = 'Cluster 2')
plt.scatter(X[y_kmeans == 2, 0], X[y_kmeans == 2, 1], s = 100, c = 'green', label = 'Cluster 3')
plt.scatter(X[y_kmeans == 3, 0], X[y_kmeans == 3, 1], s = 100, c = 'cyan', label = 'Cluster 4')
plt.scatter(X[y_kmeans == 4, 0], X[y_kmeans == 4, 1], s = 100, c = 'magenta', label = 'Cluster 5')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s = 300, c = 'yellow', label = 'Centroids')
plt.title('Clusters of customers')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.show()
```

## Apriori Algorithm


```python
dataset = pd.read_csv('data/Market_Basket_Optimisation.csv', header=None)
```

```python
dataset.head()
```

```python
transactions = []
for i in range(0, 7501):
    transactions.append([str(dataset.values[i, j]) for j in range(0, 20)])
```

```python
transactions
```

```python
from apyori import apriori
```

```python
rules = apriori(transactions, min_support = 0.003, min_confidence = 0.2, min_lift = 3, min_length = 2)
```

```python
results = list(rules)
results
```

```python
def inspect(results):
    lhs         = [tuple(result[2][0][0])[0] for result in results]
    rhs         = [tuple(result[2][0][1])[0] for result in results]
    supports    = [result[1] for result in results]
    confidences = [result[2][0][2] for result in results]
    lifts       = [result[2][0][3] for result in results]
    return list(zip(lhs, rhs, supports, confidences, lifts))
resultsinDataFrame = pd.DataFrame(inspect(results), columns = ['Left Hand Side', 'Right Hand Side', 'Support', 'Confidence', 'Lift'])
```

```python
resultsinDataFrame
```

```python
resultsinDataFrame.sort_values(by = 'Lift', ascending = False).head(10)
```

## DBSCAN

```python
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN     
```

```python
make_moons(n_samples=300, noise=0.08, random_state=42)
```

```python
# 2. Always scale your features! 
# Distance metrics like Euclidean distance are sensitive to feature scales.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
# 3. Fit DBSCAN
# eps: neighborhood radius
# min_samples: minimum points required to form a core point
dbscan = DBSCAN(eps=0.05, min_samples=5)
clusters = dbscan.fit_predict(X_scaled)
clusters
```

```python
# 4. Inspect results
# In scikit-learn, noise points are assigned the label -1
n_clusters = len(set(clusters)) - (1 if -1 in clusters else 0)
n_noise = list(clusters).count(-1)
n_clusters, n_noise
```

```python
plt.figure(figsize=(8, 5))
scatter = plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=clusters, cmap='viridis', s=40, edgecolors='k')
plt.title(f"DBSCAN Clustering (Clusters: {n_clusters}, Noise: {n_noise})")
plt.xlabel("Feature 1 (scaled)")
plt.ylabel("Feature 2 (scaled)")
plt.colorbar(scatter, label="Cluster Label (-1 = Noise)")
plt.show()
```

```python
from sklearn.cluster import KMeans

# Fit K-Means on the same scaled data
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans_clusters = kmeans.fit_predict(X_scaled)

# Plot comparison
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot K-Means
axes[0].scatter(X_scaled[:, 0], X_scaled[:, 1], c=kmeans_clusters, cmap='Accent', s=40, edgecolors='k')
axes[0].set_title("K-Means (Fails on non-spherical shapes)")

# Plot DBSCAN
axes[1].scatter(X_scaled[:, 0], X_scaled[:, 1], c=clusters, cmap='viridis', s=40, edgecolors='k')
axes[1].set_title("DBSCAN (Correctly follows density shapes)")

plt.show()
```
