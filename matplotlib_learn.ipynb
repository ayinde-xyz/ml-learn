```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]
plt.plot(x, y);
```

```python
fig = plt.figure()
ax = fig.add_subplot()
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_axes([1, 1, 1, 1])
ax.plot(x, y)
plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(x,y);
```
