# Introduction to Matplotlib

```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
plt.plot()
```

```python
plt.plot();
```

```python
plt.plot([1, 2, 3, 4]);
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]
plt.plot(x, y);
```

```python
fig = plt.figure()
ax = fig.add_subplot()
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_axes([1, 1, 1, 1])
ax.plot(x, y)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(x,y);
type(fig), type(ax)
```

```python
ax.set(title="Simple Plot", xlabel="X axis", ylabel="Y axis")
fig.savefig("images/sample-plot.png", dpi=200)
```

## Making figures with Numpy arrays

```python
x = np.linspace(0, 10, 100)
x[:10]
```

```python
fig, ax = plt.subplots()

ax.plot(x, x**2);
```

```python
fig, ax = plt.subplots()
ax.scatter(x, np.exp(x));
```

```python
fig, ax = plt.subplots()
ax.scatter(x, np.sin(x));
```

```python
nut_butter_prices = {"Apple": 10, "Cashew": 15, "Pawpaw": 20}
fig, ax = plt.subplots()
ax.bar(nut_butter_prices.keys(), height=nut_butter_prices.values());
ax.set_title("Nut Butter Prices")
ax.set_ylabel("Prices($)");
```
